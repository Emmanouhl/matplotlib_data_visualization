```python
# pyplot is matplotlib's main plotting interface (imported as plt)
import matplotlib.pyplot as plt
# Jupyter magic: display charts directly below the cell that creates them
%matplotlib inline
# MaxNLocator controls how many tick marks an axis shows (imported here but not used in this notebook)
from matplotlib.ticker import MaxNLocator
```

```python
# Data: monthly sales (in Naira) for the first six months of the year
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120000, 145000, 138000, 165000, 190000, 215000]

# Draw a black dashed line chart: months on the x-axis, sales on the y-axis
plt.plot(months, sales, color = "black", linestyle = "--")
# Add a title and axis labels so the chart is self-explanatory
plt.title("Monthly Sales Trend")
plt.xlabel( "Months")
plt.ylabel("Sales (₦)")

# Render the chart
plt.show()
```

*Customer Acquisition Comparison*

```python
# Data: customers acquired each month by two marketing campaigns
months = [1, 2, 3, 4, 5, 6]
campaign_a = [50, 65, 72, 90, 110, 125]
campaign_b = [45, 60, 68, 75, 95, 105]

# Plot both campaigns on the same axes so they can be compared directly;
# "label" is the text that will appear in the legend
plt.plot(months, campaign_a, color = "green", label = "Campaign A")
plt.plot(months, campaign_b, color = "red", label = "Campaign B")
plt.xlabel("Month")
plt.ylabel("Customers Acquired")
# Start the x-axis at month 1 and end at month 6 (removes the empty margin)
plt.xlim(1, 6)

# legend() displays the labels defined above; show() renders the chart
plt.legend()
plt.show()
```

Two Observations About the Patterns
=

1. Campaign A consistently outperforms Campaign B: Campaign A acquired more customers than Campaign B in every single month shown.
2. Both campaigns show steady growth: Both campaigns started with low acquisition numbers in Month 1 and grew consistently, with the gap between them widening slightly in the later months.

```python
# Same campaign data as before
months = [1, 2, 3, 4, 5, 6]
campaign_a = [50, 65, 72, 90, 110, 125]
campaign_b = [45, 60, 68, 75, 95, 105]

# Create a figure with 1 row and 2 columns of charts (side by side).
# fig_1 is the whole figure; axes is an array holding the two individual charts
fig_1, axes = plt.subplots(
    nrows = 1,
    ncols = 2,
    figsize = (9, 5)
)
# Give each subplot a readable name instead of using axes[0] / axes[1]
campaign_a_chart = axes[0]
campaign_b_chart = axes[1]

# Left chart: Campaign A (green line with circle markers)
campaign_a_chart.plot(months, campaign_a, color = "green", label = "Campaign A", marker = "o")
campaign_a_chart.set_title("Campaign A Performance")
campaign_a_chart.set_xlabel("Months")
campaign_a_chart.set_ylabel("Customers Acquired")
campaign_a_chart.set_xlim(1, 6)
campaign_a_chart.legend()


# Right chart: Campaign B (red line with square markers)
campaign_b_chart.plot(months, campaign_b, color = "red", label = "Campaign B", marker = "s")
campaign_b_chart.set_title("Campaign B Performance")
campaign_b_chart.set_xlabel("Months")
campaign_b_chart.set_ylabel("Customers Acquired")
campaign_b_chart.set_xlim(1, 6)
campaign_b_chart.legend()

# tight_layout() adjusts spacing so titles and labels do not overlap
fig_1.tight_layout()
plt.show()
```

*Healthcare Performance Dashboard*

```python
# Data: three hospital indicators tracked over six months.
# They have very different scales, so each gets its own chart
months = [1, 2, 3, 4, 5, 6]
patients = [420, 450, 480, 510, 495, 540]
waiting_time = [65, 62, 58, 55, 60, 49]
satisfaction = [72, 74, 76, 78, 77, 83]

# Create 3 charts stacked vertically (3 rows, 1 column) in one figure
fig, axes = plt.subplots(
    nrows = 3,
    ncols = 1,
    figsize = (6, 10)
)
# Top chart: number of patients treated each month
axes[0].plot(
    months,
    patients,
    color = "blue",
    marker = "o"
)
axes[0].set_title("Monthly Patient Volume")
axes[0].set_xlabel("Month")
axes[0].set_ylabel("Patients Treated")



# Middle chart: average waiting time in minutes
axes[1].plot(
    months,
    waiting_time,
    color = "red",
    marker = "s"
)
axes[1].set_title("Average Waiting Time")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Waiting Time (Minutes)")



# Bottom chart: patient satisfaction score (0-100 scale)
axes[2].plot(
    months,
    satisfaction,
    color = "green",
    marker = "^"
)
axes[2].set_title("Patient Satisfactory Score")
axes[2].set_xlabel("Month")
axes[2].set_ylabel("Satisfaction (0 - 100)")
    

# Prevent the stacked charts from overlapping, then display the figure
plt.tight_layout()
plt.show()
```

Report Summary:
=
The hospital's performance data over the six months shows a generally positive trend. Patient volume increased from 420 to 540, representing a 28.6% growth in demand. Despite this increased workload, average waiting time decreased significantly from 65 minutes to 49 minutes. Patient satisfaction scores mirrored this operational improvement, rising steadily from 72 to 83. This suggests the hospital successfully managed higher patient volumes without compromising wait times or patient experience.

*Bonus Challenge: Think Like a Healthcare Analyst*
=

Why these indicators should not be plotted on the same y-axis:
The variables have completely different units and scales. Patient volume is in the hundreds (420-540), waiting time is in tens of minutes (49-65), and satisfaction is on a 0-100 scale. Plotting them on one axis would compress the lines and make it impossible to distinguish the individual trends. They need separate subplots to be readable.

### Answers to the questions:

- What happened to patient volume over the six months?

It increased overall, with a slight dip in Month 5 before reaching its highest point in Month 6.

- How did average waiting time change?

It generally decreased over the period, with a minor increase in Month 5, ending at its lowest point (49 minutes) in Month 6.

- What happened to satisfaction in the final month?

It increased to its highest level (83).

- Do the patterns establish that shorter waiting times caused higher satisfaction? Why or why not?

No, they do not establish causation. They show correlation. While they moved in opposite directions (wait time down, satisfaction up), this is observational data. Other unseen factors (better staff, new equipment, seasonal effects) could be responsible for the change in satisfaction.


- What additional data would help investigate the relationship?

You would need:
  1. Individual patient-level data (to see if the same patients who waited longer were less satisfied).
  2. Staffing levels (to see if more staff caused shorter wait times).
  3. Patient demographics/acuity (were patients less sick in Month 6, making them easier to treat and more satisfied?).

*Project Title: Visualizing Business Performance with Matplotlib*

```python
# Re-import pyplot for the mini project (already imported above, so this is optional)
import matplotlib.pyplot as plt
```

```python
# Data: six months of business metrics
months = [1, 2, 3, 4, 5, 6]
revenue = [150000, 165000, 158000, 190000, 215000, 240000]
customers = [80, 92, 88, 110, 125, 140]
product_a = [45, 50, 48, 60, 68, 75]
product_b = [35, 40, 42, 45, 52, 58]

# Create a 2 x 2 grid of charts in one large figure (a dashboard layout)
fig, axes = plt.subplots(
    nrows = 2,
    ncols = 2,
    figsize = (14, 10)
)
# One overall title for the whole dashboard
fig.suptitle("Business Performance Overview (6-Month Report)", fontsize = 16, fontweight = "bold")

# Name each grid position: axes[row, column]
revenue_chart = axes[0, 0]      # top-left
customers_chart = axes[0, 1]    # top-right
products_chart = axes[1, 0]     # bottom-left
# The bottom-right slot is unused, so hide its axes to leave it blank
axes[1, 1].axis("off")

# Top-left: monthly revenue
revenue_chart.plot(
    months,
    revenue,
    color = "blue",
    marker = "o"
)
revenue_chart.set_title("Monthly Revenue")
revenue_chart.set_xlabel("Month")
revenue_chart.set_ylabel("Revenue (Naira)")

# Top-right: customer count
customers_chart.plot(
    months,
    customers,
    color = "red",
    marker = "^"
)
customers_chart.set_title("Customer Growth")
customers_chart.set_xlabel("Month")
customers_chart.set_ylabel("Number of Customers")


# Bottom-left: both products on one chart for comparison
# (Product B is dashed and uses a different marker so the lines are easy to tell apart)
products_chart.plot(months, product_a, color = "green", marker = "s", label = "Product A")
products_chart.plot(months, product_b, color = "orange", marker = "D", label = "Product B", linestyle = "--")
products_chart.set_title("Product Sales Comparison")
products_chart.set_xlabel("Month")
products_chart.set_ylabel("Units Sold")
products_chart.legend()

# Fix spacing, save the dashboard as a high-resolution PNG
# (the "images" folder must already exist), then display it
plt.tight_layout()
plt.savefig("images/business_performance_charts.png", dpi = 300, bbox_inches = "tight")
plt.show()
```

Business Report (Analysis and Interpretation)
=

## Observations Supported by Data:

1. Overall Revenue Pattern: Revenue generally increased over the six months, growing from 150,000 Naira in Month 1 to 240,000 Naira in Month 6. However, this growth was not perfectly linear; there was a slight decline in Month 3 compared to Month 2.
2. Customer Trend vs. Revenue Trend: The customer count followed a similar upward trajectory, growing from 80 to 140 customers. The slight dip in Month 3 is mirrored in both revenue and customer numbers, suggesting that the revenue dip was likely driven by fewer customers rather than a change in spending per customer.
3. Product Performance: Both Product A and Product B show consistent growth. Product A maintained higher monthly sales than Product B in every month, ending at 75 units compared to Product B's 58 units in Month 6.

## Interpretation (What the data does NOT prove):

- Causation vs. Correlation: The data shows that revenue and customers move together, but it does not prove that acquiring more customers caused the revenue to rise. It is possible that existing customers simply spent more, or that prices were increased.
- Month 3 Dip: The data shows a dip in Month 3, but without additional context, we cannot know why. It could be due to a seasonal slowdown, a competitor's promotion, or a temporary supply issue. Do not assume a specific cause without supporting evidence.
- Additional Information Needed: To better understand the business, the owner should also track:
  · Marketing Spend: To see if increased spending correlates with customer growth.
  · Average Order Value (AOV): To see if customers are spending more or less per transaction over time.
  · Customer Retention Rate: To see if the business is keeping the new customers it acquires.
